### Pandas - Library for Data visualization and Data manipulation

```python
import pandas as pd
#from tabulate import tabulate
order_details = pd.read_table(filepath_or_buffer="data.tsv.txt", sep="\t")
order_details
```

### Tasks

#### Find out the top 5 selling items individual revenue it made

```python
top5_items = order_details.groupby(by='item_name')['quantity'].sum().sort_values(ascending = False).head(5)
top5_items
```

#### Total revenue of the top 5 items sold

```python
order_details['item_price'] = order_details['item_price'].str.replace('$', '').str.strip().astype(float)
top5_profitable_items = order_details.groupby(by =['item_name', 'quantity'])['item_price'].sum().sort_values(ascending=False).head(5)
top5_profitable_items
```
