#### Write a program to take name and age from user and print a greeting message

```python
name = input("Please enter your name")
age = int(input("Please enter your age"))

if(age>=18):
    print(f"Hi, {name} your age is {age}. So you can cast your vote for this upcoming election")
else:
     print(f"Hi, {name} your age is {age}. Voters must be 18 years or older")
```

#### Create a list of numbers and print only even number

```python
list1 = [1,2,3,5,90,80,67,78,45,29,9,990,70,28,12,34,35]
output1 = []

print(f'Number of elements present in list : {len(list1)}')
for i in range(1, len(list1)):
    if(list1[i] % 2 == 0):
        output1.append(list1[i])

print(f'Number of even elements present in list : {len(output1)}')
print(f'Even number array : {output1}')
```

#### Exception Handling - Divison by zero


```python
input1 = int(input("Enter dividend number :"))
input2 = int(input("Enter divisor number :"))

try: 
    output = input1/input2
    print(f'Divison output : {output}')
except ZeroDivisionError:
    print("Error: numbers cant be divided by zero")
```

#### Given a string count how many vowel it has

```python
vowels = ('a','e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U')

input1 = input("Enter a string input ")
count = 0
for char in input1:
    if(char in vowels):
        count+=1
print(f'Number of vowels present : {count}')
```

#### Reverse a list without inbuilt function

```python
input1 = [1,2,3,4,5]
reversed_list = []

for i in range(len(input1) - 1, -1, -1): 
    reversed_list.append(input1[i])

print(reversed_list)
    
```
