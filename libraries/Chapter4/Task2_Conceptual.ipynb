### When would you use list over a set or a tuple?

List Features: 
 -> Supports Indexing and Slicing
 -> Ordered Data structure, Allows duplication, Mutable
 -> Has more inbuilt methods like pop, append, sort
 -> Dynamic  Memory allocation

Set Features:
 -> It doesnt allow duplication, Non ordered data structure
 -> It doesnt allow indexing or slicing 
 -> It is mutable

Tuple Features:
 -> Python's default 1D Data structure
 -> Supports Indexing and Slicing 
 -> immutable and fast compared to list
 -> Have few inbuilt methods such as count(), index()

If the requirement needs to store data in 1 dimensional structure which needs to be ordered, allows duplication and change of values in data in future we can go for list structure

### Exceptional handling and example

In our code, if we expect some failure during the runtime it might cause program termination we can handle those with exceptional handling. It is used to intercept the error and can continue running instead of abruptly crashing the program executions. Methods used in exceptional handling try, except and finally 

The code which we suspect that will cause error can be inside try block and it raise the exception, which can be handled in except block we can print relevant error message useful for debugging.

Eventhough there is an exception, my code needs to perform any action mandatorily we can write that inside Finally block. Whether the exception is handled or not the finally block will execute. (ex: Closing the database)

```python
## Handling division by Zero

inp1 = int(input("Enter dividend number: "))
inp2 = int(input("enter divisor  number: "))

try:
    result = inp1/inp2
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero")

finally:
    print("Execution completed")
```

### Series and DataFrames

In Pandas library, 2 type of data structures are there:
-> Series - 1 dimensional datastructure labelled with index
-> DataFrame - 2 dimensioanl datastructure



```python
print("############ Series ##################")
import pandas as pd

names = ['Meenakshi', 'Dheerthi', 'Kabilan', 'Raji', 'Kamal']

name_series = pd.Series(names, index = [0,1,2,3,4])
print("\n", name_series)

print("\n ############### Pandas #############")

marks = {'Name' : ['Veena', 'Viyaan', 'Aishu', 'Mani', 'Rishi'],
          'Marks' : [90, 98, 100, 88, 78]}

student_marks = pd.DataFrame(marks)

print("\n", student_marks)
```

### What are the different ways of handling missing values in dataset

Step 1: Check for Duplicates in Data set and remove the duplicate entries 
        Method - dataframe.drop_duplicates()

Step 2: Check the null values in dataset 
        Method - dataframe.isna() , dataframe.isnull()
        to get the summarized count value - df.isnull().sum()

Step 3: Get the approval from client and remove the null entries in data set
        Method - dataframe.dropna() [to remove observations], dataframe.dropna(axis=1) [to remove features]

Step 4 : Can fill the non important column like name, data, id with constant values 
         df.fillna(0) and df.fillna('unknown')

         Can fill with Mean, Median, Mode
         df['column1'].fillna(df['column1'].mean())
         df['column1'].fillna(df['column1'].median())
         df['column1'].fillna(df['column1'].mode()[3]) -> the index number represents most frequent value in the column

### Matplotlib and Seaborn

#### Matplotlib

 -> Role: It provides total control over every single element of a chart (axes, labels, fonts, colors).

 -> Best for: Creating highly customised, complex plots or fine-tuning the small details of a graph.

 -> The Downside: It often requires a lot of code to make a basic chart look "modern" or attractive.
 
#### Seaborn (Built on top of Matplotlib)

 -> Role: It is designed for statistical exploration. It can create complex visualisations (like heatmaps or distribution plots) in just one or two lines of code.

 -> Best for: Exploring relationships between variables, handling categorical data, and creating "publication-ready" charts with beautiful default themes.

 -> The Strength: It automatically handles things like legends and color mapping that you’d have to code manually in Matplotlib.

### loc and iloc difference 

loc and iloc are used to filter / slice data from dataframe

 -> loc - label based
    ex: df.loc('meenakshi', 'Age') where selects the age data from observation value 'meenakshi' 

        df.loc('abi':'deena", 'Age') where fetches the  age value from abi to deena observation 

 -> iloc - index based  ex: df.iloc(0,0)
    ex: df.iloc(0,0) slice the value where observation index 0 and feature index 0

        df.iloc(0:5, 0:5) slice the value from 0th to 5th index in observation and feature 0 to 5
