### Abstraction
Hides the internal implementation and shows only essential features to the end user.
Abstraction allows to define common interface and all the subclasses must implement the methods

```python
from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    @abstractmethod
    def payby(self, amount):
        pass


class UPIPayment(PaymentProcessor):
    def payby(self, amount):
        print(f" Rs.{amount} is paid via UPI")

class CardPayment(PaymentProcessor):
    def payby(self, amount):
        print(f" Rs.{amount} is paid via card")

upi = UPIPayment()
upi.payby(1000)

card = CardPayment()
card.payby(2000)
```

## Example 2

```python
from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class GoKwik(PaymentProcessor):
    app = 'gokwik';
    def pay(self, amount):
        print(f"RS. {amount} paid via {self.app}")

class RazorPay(PaymentProcessor):
    app = 'RazorPay'
    def pay(self, amount):
        print(f"RS. {amount} paid via {self.app}")




class Purchase:
    def __init__(self, gateway):
        self.gateway = gateway

    def confirmPayment(self, amount):
        self.gateway.pay(amount)
        



gokwik = GoKwik()
razorPay = RazorPay()

obj1 = Purchase(gokwik)
obj1.confirmPayment(1000)

obj2 = Purchase(razorPay)
obj2.confirmPayment(2000)
```
