## Method Overloading: 
 Python doesnt support overloading natively we have to use none /  variable args
 Method overloading in other languages
 sum(a, b) and sum(a,b,c) method with same name but different type or number of arguments 
 this wont work in  python 

```python
class Calc:
    def add(self, a, b, c=None):
        if c == None: 
            return a+b
        return a + b + c


calc = Calc()
print(calc.add(10,12))
print(calc.add(10,12,10))
```

```python
class Calc:
    def mul(self , *args):
        if not args:
            return 0
        result = 1
        for i in args:
            result *= i

        return result

calc = Calc()
print(calc.mul())
print(calc.mul(1,2))
print(calc.mul(1,2,10))
print(calc.mul(3,2,9, 5,6))
```
