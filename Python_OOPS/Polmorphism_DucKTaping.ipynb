## DuckTaping :
The principle is 'if it walks like a duck, quacks like a duck  it is a duck"
Meaning: Object suitability is determined by its methods or properties rather than its types


```python
class Laptop:
   def code(self, ide):
       ide.execute()

class Pycharm:
    def execute(self):
        print("Step 1 - Compiles the code")
        print("Step 2 - Runs the code")

class CustomEditor:
    def execute(self):
        print("Step 1 - Spell check")
        print("Step 2 - Convention check")
        print("Step 3 - Compiles the code")
        print("Step 4 - Runs the code")

obj1 = Pycharm()
obj1.execute()
print ("----------------------")

obj2 = CustomEditor()
obj2.execute()
```
