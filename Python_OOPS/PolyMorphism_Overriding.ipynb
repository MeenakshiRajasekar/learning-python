## Method overriding: 
When super class and sub class  have same method in both classes, sub class method will override the super class method


```python
class Employee:
    def __init__(self, name, id_no):
        self.name = name
        self.id_no = id_no

    def __login__(self):
        print("Employee should login by 9 am")

    def __taskAction__(self):
        print("Employee can view the task")

class TeamMember(Employee):
    def __taskAction__(self):
        print("Employee can view the task")
        print("Employee can submit the task")

class Manager(Employee):
    def __taskAction__(self):
        print("Manager can view the task")
        print("Manager can Assign the task")


manager =  Manager('Meenakshi', 20)
manager.__taskAction__()
manager.__login__()
print("------------------------------")
team = TeamMember('Meena',100)
team.__taskAction__()
```
