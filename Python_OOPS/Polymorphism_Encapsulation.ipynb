### Encapsulation 

```python
class Employee_Salary:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

    def get_details(self):
        return self.__salary

    def update_salary(self, amount):
        if(amount>1000):
            self.__salary += amount
        else:
            print("amount is invalid")


employee_salary =  Employee_Salary('Meenakshi', 20000)

print("befor update ", employee_salary.get_details())
employee_salary.update_salary(10000)
print("after update ",employee_salary.get_details())
```

```python
class Bank:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def get_balanceDetails(self):
        return self.__balance

    def deposit(self, amount):
        if amount <=0:
            print ("Amount is not sufficient")
        else:
            self.__balance += amount

    def cash_withdrawal(self, amount):
        if amount > self.__salary:
            print("Insufficient funds")

        else:
            self.__balance -= amount

print("Actual balance")
bank = Bank('Meenakshi', 70000)
print(bank.get_balanceDetails())
bank.deposit(10000)
print("After a deposit")
print(bank.get_balanceDetails())

bank.cash_withdrawal
print("After a withdrawal")
```
