```python
class Employee:
    def __init__(self, name):
        self.name = name
        
    def login(self):
        print(f"{self.name} has logged in")

    def viewTask(self):
        print(f"{self.name} is viewing the task")

class TeamMember(Employee):
    def submitTask(self):
        print(f"{self.name} has submitted the task")

class Manager(Employee):
    def assignTask(self):
        print(f"{self.name} has assigned the task")

```

```python
meena = TeamMember("Meenakshi")
meena.viewTask()
```
