## Operator overloading:
1. It allows to change or define the in-built properties like ('+', '-', '*', '='....)
2. Python doesnt know how to add two custom objects. Operator overloading allows us to define the logic explicitly.

#### + operator uses __add__(self, other)

```python
class Income:
    def __init__(self, name , salary):
        self.name = name
        self.salary = salary

        
    def __add__(self, other):
        total = self.salary + other.salary
        return total

obj1 = Income('Meena', 25000)
obj2 = Income('Kabil', 30000)

print(obj1 + obj2)
```

### - operator uses __sub__(self, other)

```python
class Student:
    def __init__(self, name, mark):
        self.name = name
        self.mark = mark

    def __sub__(self, other):
        mark_diff = self.mark - other.mark
        return mark_diff

obj1 = Student('Meenakshi', 100)
obj2 = Student('Aishu', 92)
mark_diff = obj1 - obj2

print(f"The mark diff is  : {mark_diff}" )
```

```python
class Laptop:
    def __init__(self, brand, ram):
        self.brand = brand
        self.ram = ram

    def __sub__(self, ram_to_add):
        new_ram = self.ram - ram_to_add
        return Laptop(self.brand, new_ram)
        
    def show(self):
        print(f"{self.brand} laptop ram is downgraded to {self.ram}")

obj1 = Laptop('HP', 32)

downgraded_ram = obj1 - 16
downgraded_ram.show()
```

### * to multiply the own objects __mul__(self, other)


```python
class Laptop:
    def __init__(self, brand, ram):
        self.brand = brand
        self.ram = ram

    def __mul__(self, upgrade):
        new_ram = self.ram * upgrade
        return Laptop(self.brand, new_ram)

    def show(self):
        print(f" {self.brand} laptop ram is now upgraded to {self.ram}")

obj1 = Laptop('HP', 8)

upgrade_ram = obj1  * 6

upgrade_ram.show()
```
